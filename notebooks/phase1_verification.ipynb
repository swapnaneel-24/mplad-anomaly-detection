```python
import hashlib
import os
import pandas as pd

# Define relative path from notebooks/ to data/raw/
DATA_PATH = os.path.join("..", "data", "raw", "creditcard.csv")

print(f"Loading dataset from: {DATA_PATH}")
df = pd.read_csv(DATA_PATH)

# --- 1. Structural Verification ---
rows, cols = df.shape
print(f"[*] Row count: {rows} | Expected: 284807")
print(f"[*] Column count: {cols} | Expected: 31")

assert rows == 284807, f"FATAL: Row count mismatch. Found {rows}"
assert cols == 31, f"FATAL: Column count mismatch. Found {cols}"

# --- 2. Class Distribution Verification ---
positive_count = int(df["Class"].sum())
negative_count = int((df["Class"] == 0).sum())
fraud_ratio = (positive_count / rows) * 100

print(f"[*] Positive class (Fraud): {positive_count} | Expected: 492")
print(f"[*] Negative class (Genuine): {negative_count} | Expected: 284315")
print(f"[*] Class ratio (Fraud %): {fraud_ratio:.4f}%")

assert positive_count == 492, f"FATAL: Fraud count mismatch. Found {positive_count}"
assert negative_count == 284315, f"FATAL: Genuine count mismatch. Found {negative_count}"

# --- 3. Missing Value Verification ---
null_count = int(df.isnull().sum().sum())
print(f"[*] Total missing values: {null_count} | Expected: 0")
assert null_count == 0, f"FATAL: Found {null_count} missing values."

# --- 4. Cryptographic Hash (SHA-256) ---
sha256_hash = hashlib.sha256()
with open(DATA_PATH, "rb") as f:
    for byte_block in iter(lambda: f.read(65536), b""):
        sha256_hash.update(byte_block)

file_checksum = sha256_hash.hexdigest()
print(f"\n[+] Verification Passed.")
print(f"[+] SHA-256 Checksum: {file_checksum}")
```
